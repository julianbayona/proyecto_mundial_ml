# Fase 1 - Carga y cruce de datos

Este notebook llama la funcion reutilizable `load_world_cup_data()`, revisa las primeras filas, valida que exista la columna `stage` y guarda el resultado crudo cruzado. El alcance queda limitado hasta 2014 porque `WorldCupMatches.csv` no incluye las ediciones 2018 y 2022.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path("..") if Path.cwd().name == "notebooks" else Path(".")
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.data_loader import load_world_cup_data
```

```python
world_cup_matches = load_world_cup_data(project_root=PROJECT_ROOT)
print(world_cup_matches.shape)
world_cup_matches.head()
```

```python
assert "stage" in world_cup_matches.columns, "La columna stage no esta presente."
print("Columna stage presente.")
print(world_cup_matches["stage"].isna().sum(), "partidos sin stage")
```

```python
output_path = PROJECT_ROOT / "data" / "processed" / "world_cup_matches_raw.csv"
output_path.parent.mkdir(parents=True, exist_ok=True)
world_cup_matches.to_csv(output_path, index=False)
print(f"Dataset guardado en: {output_path}")
```
