```python
from pathlib import Path
import sys

PROJECT_ROOT = Path("..") if Path.cwd().name == "notebooks" else Path(".")
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.preprocessing import clean_world_cup_data
```

```python
clean_df = clean_world_cup_data(project_root=PROJECT_ROOT)
clean_df.shape
```

```python
print(clean_df["stage"].value_counts())
print(clean_df["is_group_stage"].value_counts())
assert "is_group_stage" in clean_df.columns
assert "host_continent" in clean_df.columns
assert clean_df["host_continent"].isna().sum() == 0
```
