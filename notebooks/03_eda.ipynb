## Conclusiones preliminares

- El dataset limpio contiene 836 partidos entre 1930 y 2014. La cantidad de partidos por edicion crece con los cambios historicos de formato, hasta llegar a 64 partidos en las ediciones modernas.
- El promedio de goles por partido es mas alto en ediciones antiguas como 1954 y mas bajo en torneos modernos como 1990, lo que sugiere cambios historicos en estilo de juego y formato competitivo.
- La variable temporal de avance queda razonablemente balanceada para exploracion inicial: 199 selecciones-edicion avanzan y 225 no avanzan.
- Las selecciones de la misma region continental de la sede muestran una tasa de avance mayor en esta muestra descriptiva: 0.521 frente a 0.436 para el resto. Esto no implica causalidad, pero justifica analizar el efecto regional en fases posteriores.
- Las variables `goal_difference` y `total_goals` se usan solo para EDA. No deben incluirse como features del modelo supervisado porque dependen de resultados del torneo actual.
